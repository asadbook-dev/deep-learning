# OpenCV

```python
import numpy as np
import cv2 as cv

img = cv.imread('dog.jpg')
assert img is not None, "file could not be read, check with os.path.exists()" # rasm to'g'ri o'qilganmi yoki yo'q shuni tekshiradi. File topilmasa xatolik bermaydi None qaytaradi. File yo'q bo'lsa img = None
```

```python
cat_image = cv.imread('cat.jpeg')
assert cat_image is not None, "file could not be read, check with os.path.exists()"

print("Type:", type(cat_image))
print("Shape (height, width, channels):", cat_image.shape)
print("Size :", cat_image.size)
print("Data type :", cat_image.dtype)
```

```python
cat_image
```

```python
px = cat_image[10, 10] # Rasmdagi bitta pixelning qiymatini oladi: 10-qator, 10-ustundagi pixel.
print(px)
# 183 => Blue, 168 => Green, 159 => Red.    ---- (BGR)  -----   not (RGB)
```

```python
px_b = cat_image[10, 10, 0]
px_g = cat_image[10, 10, 1]
px_r = cat_image[10, 10, 2]
print("Blue: ", px_b)
print("Green: ", px_g)
print("Red: ", px_r)
```

```python
cat_image[10, 10] = [255, 255, 255] # Sonlarni ham o'zgartira olamiz
print(cat_image[10, 10])
```

# PIL

```python
from PIL import Image
with Image.open("cat.jpeg") as cat: # Rasmni ochadi va cat deb nomlaydi. with ishi tugagach faylni avtomatik yopadi (xotira tozalanadi).
    cat.rotate(45).show() # rotate(45): rasmni soat strelkasiga teskari yo'nalishda 45° buradi

print("=== PIL ===")
print("Type:", type(cat))
print("Format:", cat.format)
print("Mode:", cat.mode)
print("Size (width, height):", cat.size)
```
