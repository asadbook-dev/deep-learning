# Learning Rate (RL) ni tanlash

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# Data
X = torch.randn(200,1)
y = 4*X + 1
```

```python
X
```

```python
def train_model(lr):
    model = nn.Linear(1,1)
    optimizer = optim.SGD(model.parameters(), lr=lr)
    criterion = nn.MSELoss()
    losses = []
    
    for epoch in range(50):
        optimizer.zero_grad()
        output = model(X)
        loss = criterion(output, y)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())
    
    return losses
```

```python
lr_small = train_model(0.0001)
lr_good = train_model(0.01)
lr_large = train_model(1)
```

```python
plt.plot(lr_small, label="LR=0.0001")
plt.plot(lr_good, label="LR=0.01")
plt.plot(lr_large, label="LR=1")
plt.legend()
plt.title("Learning Rate Comparison")
plt.show()
```

```python
model = nn.Linear(1,1)
optimizer = optim.SGD(model.parameters(), lr=0.1)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)

criterion = nn.MSELoss()

for epoch in range(30):
    optimizer.zero_grad()
    output = model(X)
    loss = criterion(output, y)
    loss.backward()
    optimizer.step()
    scheduler.step()
    
    print("Epoch:", epoch, "LR:", optimizer.param_groups[0]['lr'])
```
