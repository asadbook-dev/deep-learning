```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("cat.jpeg")

def show_img(image, title):
    plt.imshow(image)
    plt.title(title)
    plt.axis("off")
    plt.show()

show_img(img, "Original cat image")
```

```python
output_dir = Path("augmented_images")
output_dir.mkdir(parents=True, exist_ok=True)

def save_image(image, name):
    output_path = output_dir / f"{name}.jpeg"
    image.save(output_path)
```

# Horizontal Flip

```python
from torchvision import transforms

transform = transforms.RandomHorizontalFlip()

new_img = transform(img)



save_image(new_img, "horizonal_flip")
show_img(new_img, "Horizontal Flip")
```

# Vertical Flip

```python

transform = transforms.RandomVerticalFlip()

new_img = transform(img)

save_image(new_img, "vertical_flip")
show_img(new_img, "Vertical Flip")
```

# Rotation

```python
transform = transforms.RandomRotation(degrees=110)
new_img = transform(img)

save_image(new_img, "rotation")
show_img(new_img, "Rotation")
```

# Crop

```python
transform = transforms.RandomCrop(size=(250, 230))
new_img = transform(img)

save_image(new_img, "crop")
show_img(new_img, "Crop")
```

# Random Affine

```python

transform = transforms.RandomAffine(
    degrees=30, # rotation (-30,30)
    translate=(0.1, 0.1),   # horizontally and verticaly
    scale=(0.8, 1.9),  # zoom
    shear=10 #(-10,10)  # qiyalashtirish(yotkizish)
)
new_img = transform(img)


save_image(new_img, "affine")
show_img(new_img, "Affine")
```

# GrayScale

```python
transform = transforms.RandomGrayscale(p=0.8)
new_img = transform(img)

save_image(new_img, "grayscale")
show_img(new_img, "GrayScale")
```

# Combine Augmentation

```python
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(30),
    transforms.ColorJitter(brightness=0.3, contrast=0.3),
    transforms.RandomCrop((150 , 150))
])

new_img = transform(img)

save_image(new_img, "combine_augmentation")
show_img(new_img, "Combine Augmentation")
```
