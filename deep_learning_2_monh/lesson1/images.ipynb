# RGB

```python
import numpy as np
import matplotlib.pyplot as plt

image = np.random.randint(0,255,(20, 20, 3))

image

plt.imshow(image)
plt.title("Random RGB Image")
plt.show()
```

```python
image.shape 
# 20 pixelga 20 pixel image
# height => 20, weight => 20, channels => 3 (R, G, B)
```

```python
red = image[:,:,0] # => hamma qatorni, hamma ustunni olib 0-index yani R channelni oladi. qolgan 2ta G, B channel o'chadi. Ular yo'q holatdagi image qanday ko'rinishi
green = image[:,:,1] # => 1-index => faqat Green channelni oladi qolganini olmaydi
blue = image[:,:,2] # => 2-index => faqat Blue channelni oladi qolganini olmaydi

plt.imshow(red, cmap='Reds')
plt.title("Red Channel")
plt.show()
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,4,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(red, cmap='Reds')
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(green, cmap='Greens')
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(blue, cmap='Blues')
plt.title("Blue Channel")
plt.axis("off")

plt.show()
```

```python
# 1-si original imagemiz.
# 2-si Green va Blue channellar ni olib tashlagan holatidagisi
# 3-si Red va Blue channellar yoq holati
# 4-si Red va Green channellar yo'q holati
# 2, 3, 4 rasm birlashib 1-dagi raasmni hosil qilayotgan edi
```

```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

img = Image.open("image.jpeg")


img_array = np.array(img)

red = img_array[:, :, 0]
green = img_array[:, :, 1]
blue = img_array[:, :, 2]


print("Image type:", type(img))
print("Array type:", type(img_array))
print("Image shape:", img_array.shape)


plt.imshow(img_array)
plt.title("Original Image")
plt.axis("off")
plt.show()
```
