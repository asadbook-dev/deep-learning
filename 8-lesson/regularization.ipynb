```python
import numpy as np

df['Price'] = df['Price'].replace('Ask For Price', np.nan)

df['Price'] = (
    df['Price']
    .astype(str)
    .str.replace(',', '', regex=False)
)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

df['kms_driven'] = (
    df['kms_driven']
    .astype(str)
    .str.replace(',', '', regex=False)     
    .str.replace('kms', '', regex=False)   
    .str.strip()                            
)

df['kms_driven'] = pd.to_numeric(df['kms_driven'], errors='coerce')

df['year'] = pd.to_numeric(df['year'], errors='coerce')
```

```python
from sklearn.preprocessing import LabelEncoder


TARGET_COL = "Price"
BATCH_SIZE = 64
EPOCHS = 7
learning_rates = [0.1, 0.01, 0.001, 0.0001, 0.00001]

x = df.drop(columns=[TARGET_COL]) # Features

y = df[TARGET_COL] # Target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Encoding categorical features using LabelEncoder
def encode_train(df):
    df = df.copy()
    encoders = {}
    for col in df.columns:
        if not pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].astype(str)
            encoder = LabelEncoder()
            df[col] = encoder.fit_transform(df[col])
            encoders[col] = encoder
    return df, encoders

def encode_test(df, encoders):
    df = df.copy()
    for col, encoder in encoders.items():
        df[col] = df[col].astype(str)
        known = set(encoder.classes_)
        df[col] = df[col].apply(lambda x: encoder.transform([x])[0] if x in known else -1)
    return df

encoded_xtrain, encoders = encode_train(x_train)   # train
encoded_xtest = encode_test(x_test, encoders)       # test
```

# SML

```python
x__train = encoded_xtrain.copy()
x__test = encoded_xtest.copy()
y__train = y_train.copy()
y__test = y_test.copy()
```

```python
# Scaling
from sklearn.preprocessing import MinMaxScaler
def scale_train(df):
    scalers = {}

    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            scaler = MinMaxScaler()
            df[col] = scaler.fit_transform(df[[col]])
            scalers[col] = scaler
    return df, scalers

def scale_test(df, scalers):
    for col in df.columns:
        if col in scalers:
            scaler = scalers[col]
            df[col] = scaler.transform(df[[col]])
    return df

scaled__xtrain, scalers = scale_train(x__train)   # train
scaled__xtest = scale_test(x__test, scalers)      # test
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

lr_model = RandomForestRegressor()
lr_model.fit(scaled__xtrain, y__train)
y_pred = lr_model.predict(scaled__xtest)

print("R2 score of Random Forest model on test data:", r2_score(y__test, y_pred))
```

```python

# Pandas to Numpy
x_train = encoded_xtrain.to_numpy()
x_test = encoded_xtest.to_numpy()
y_train_arr = y_train.to_numpy().reshape(-1, 1)
y_test_arr = y_test.to_numpy().reshape(-1, 1)

# Scaling the features and target variable
from sklearn.preprocessing import StandardScaler

x_scaler = MinMaxScaler()
scaled_xtrain = x_scaler.fit_transform(encoded_xtrain)
scaled_xtest  = x_scaler.transform(encoded_xtest)

y_scaler = MinMaxScaler()
scaled_ytrain = y_scaler.fit_transform(y_train_arr)
scaled_ytest  = y_scaler.transform(y_test_arr)

# Numpy to Torch Tensor
x_train_tensor = torch.tensor(scaled_xtrain, dtype=torch.float32)
x_test_tensor = torch.tensor(scaled_xtest, dtype=torch.float32)

y_train_tensor = torch.tensor(scaled_ytrain, dtype=torch.float32)
y_test_tensor = torch.tensor(scaled_ytest, dtype=torch.float32)
```

```python
print("X Train Tensor:", x_train_tensor.shape)
print("Y Train Tensor:", y_train_tensor.shape)
print("X Test Tensor:", x_test_tensor.shape)
print("Y Test Tensor:", y_test_tensor.shape)
```

```python
# TensorDataset and DataLoader
from torch.utils.data import TensorDataset, DataLoader, random_split

dataset = TensorDataset(
    x_train_tensor,
    y_train_tensor
)

train_size = int(0.9 * len(dataset))
val_size = len(dataset) - train_size

train_dataset, val_dataset = random_split(
    dataset,
    [train_size, val_size]
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_dataset = TensorDataset(
    x_test_tensor,
    y_test_tensor
)
test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)
```

```python
# Model
from torch import nn, optim



torch.manual_seed(0) # random seed for reproducibility
input_dim = x_train_tensor.shape[1] # features
hidden_dim = 64 # hidden units
output_dim = 1 # output units


def train_model(ALPHA):
    model = nn.Sequential(
        nn.Linear(input_dim, hidden_dim),   #Intput
        nn.ReLU(),   #Hidden 
        nn.Linear(hidden_dim, output_dim)   #Output
    )
    loss_fn = nn.MSELoss()   
    opt = optim.Adam(model.parameters(), lr=ALPHA)
    best_val_loss = float("inf")
    best_val_r2 = float("-inf")

    for epoch in range(1, EPOCHS + 1):
        model.train() 
        total_loss = 0.0  

        for bx, by in train_loader:
            pred = model(bx) 
            loss = loss_fn(pred, by) 
            opt.zero_grad() 
            loss.backward() 
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step() 
            total_loss += loss.item() 
        train_loss = total_loss / len(train_loader) 

        # Validation
        model.eval()

        total_val_loss = 0.0

        all_preds = []
        all_targets = []

        with torch.no_grad():
            for bx, by in val_loader:

                pred = model(bx)
                loss = loss_fn(pred, by)
                total_val_loss += loss.item()   
                all_preds.extend(pred.squeeze().cpu().numpy())
                all_targets.extend(by.squeeze().cpu().numpy())

        val_loss = total_val_loss / len(val_loader)
        val_r2 = r2_score(all_targets, all_preds)

        print(
            f"LR: {ALPHA} | "
            f"Epoch: {epoch} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val R2: {val_r2:.4f}"
        )

        if val_loss < best_val_loss:
            best_val_loss = val_loss

        if best_val_r2 < val_r2:
            best_val_r2 = val_r2

    return best_val_loss, best_val_r2
```

# Learning rate (experiment)

```python
results = {}

for lr in learning_rates:

    print(f"\n===== Learning Rate: {lr} =====")

    best_val_loss, best_val_r2 = train_model(lr)

    results[lr] = (best_val_loss, best_val_r2)

print(results)
```

# Learning Rate (Scheduler)

```python
# Model
from torch import nn, optim



torch.manual_seed(0) # random seed for reproducibility
input_dim = x_train_tensor.shape[1] # features
hidden_dim = 64 # hidden units
output_dim = 1 # output units


def train_model(ALPHA):
    model = nn.Sequential(
        nn.Linear(input_dim, hidden_dim),   #Intput
        nn.ReLU(),   #Hidden 
        nn.Linear(hidden_dim, output_dim)   #Output
    )
    loss_fn = nn.MSELoss()   
    opt = optim.Adam(model.parameters(), lr=ALPHA)
    scheduler = optim.lr_scheduler.StepLR(opt, step_size=10, gamma=0.1)

    best_val_loss = float("inf")
    best_val_r2 = float("-inf")

    for epoch in range(1, EPOCHS + 1):
        model.train() 
        total_loss = 0.0  

        for bx, by in train_loader:
            pred = model(bx) 
            loss = loss_fn(pred, by) 
            opt.zero_grad() 
            loss.backward() 
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step() 
            total_loss += loss.item() 
        train_loss = total_loss / len(train_loader) 

        # Validation
        model.eval()

        total_val_loss = 0.0

        all_preds = []
        all_targets = []

        with torch.no_grad():
            for bx, by in val_loader:

                pred = model(bx)
                loss = loss_fn(pred, by)
                total_val_loss += loss.item()   
                all_preds.extend(pred.squeeze().cpu().numpy())
                all_targets.extend(by.squeeze().cpu().numpy())

        val_loss = total_val_loss / len(val_loader)
        val_r2 = r2_score(all_targets, all_preds)
        scheduler.step(val_loss)

        print(
            f"LR: {opt.param_groups[0]["lr"]} | "
            f"Epoch: {epoch} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val R2: {val_r2:.4f}"
        )

        if val_loss < best_val_loss:
            best_val_loss = val_loss

        if best_val_r2 < val_r2:
            best_val_r2 = val_r2

    return best_val_loss, best_val_r2
```

# BatchNorm regularization
