```python
#ImageEnhance - Rasm sifatini bitta son (factor) bilan o'zgartiradi.
# 1. Brightness (yorqinlik): har pixelni f ga ko'paytiradi.
# 2. Contrast: pixellarni o'rtacha (m) dan uzoqlashtiradi yoki yaqinlashtiradi.
# 3. Color (to'yinganlik, saturation): rangni grayscale (g) dan uzoqlashtiradi.
# 4. Sharpness: qirralarni aniqroq (f > 1) yoki xiraroq (f < 1) qiladi.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance

img = Image.open("../dataset/train/phones/iphone.jpg").convert("RGB")

plt.imshow(img)
plt.title("Original")
plt.axis("off")
plt.show()
```

# Brightness

```python
# natija 255 dan oshsa 255 qilinadi, 0 dan kichik bo'lsa 0.

brightness = ImageEnhance.Brightness
factor = 0.3 

dark_img = brightness(img).enhance(factor)

plt.imshow(dark_img)
plt.title("Brightness")
plt.axis("off")
plt.show()

```

```python
# pixellarni o'rtacha qiymatdan uzoqlashtiramiz (f > 1) yoki yaqinlashtiramiz (f < 1). Yorqin joy yanada yorqin, qorong'i joy yanada qorong'i bo'ladi.
contrast = ImageEnhance.Contrast
factor = 0.5
contrast_img = contrast(img).enhance(factor)

plt.imshow(contrast_img)
plt.title("Contrast")
plt.axis("off")
plt.show()
```

```python
# rangni kulrang holatdan uzoqlashtiramiz ($f > 1$, ranglar jonli) yoki unga yaqinlashtiramiz ($f < 1$, ranglar so'nadi).

factor = 2.0
color_img = ImageEnhance.Color(img).enhance(factor)

plt.imshow(color_img)
plt.title("Color")
plt.axis("off")
plt.show()
```

```python
# Sharpness = rasmdagi qirralar (chegaralar) qanchalik aniq ko'rinishini sozlaydi.
# PIL avval rasmning xiralashtirilgan nusxasini ($b$) yasaydi, keyin asl rasm ($x$) bilan aralashtiradi:
# asl rasm bilan xira nusxasi orasidagi farq = qirralar. Factor shu farqni kuchaytiradi yoki kamaytiradi.

factor = 2.0
sharp_img = ImageEnhance.Sharpness(img).enhance(factor)

plt.imshow(sharp_img)
plt.title("Sharp")
plt.axis("off")
plt.show()
```

```python


enhancers = {
    "Brightness": ImageEnhance.Brightness,
    "Contrast":   ImageEnhance.Contrast,
    "Color":      ImageEnhance.Color,
    "Sharpness":  ImageEnhance.Sharpness,
}
factors = [0.3, 1.0, 2.0]

fig, axes = plt.subplots(4, 3, figsize=(10, 12))

for row, (name, cls) in enumerate(enhancers.items()):
    for col, f in enumerate(factors):
        out = cls(img).enhance(f)
        mean_val = np.array(out).mean()          # barcha pixel o'rtachasi
        axes[row, col].imshow(out)
        axes[row, col].set_title(f"{name} f={f}\nmean={mean_val:.0f}")
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Brightness	umumiy yorug'lik
# Contrast	    yorqin-qorong'i farqi
# Color	        rang jonliligi
# Sharpness	    qirralar aniqligi
```
